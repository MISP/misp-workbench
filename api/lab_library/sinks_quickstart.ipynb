# Send indicators to a sink

Shows how to push indicators from a notebook to an outbound sink (Splunk HEC, GELF, syslog/CEF or webhook)
configured under **internals → sinks**. Sending needs the `sinks:send` scope on your role.

Fork this notebook to a personal copy before running.

> **Run the cells in order**, or press **Run all**. Each notebook gets its own kernel, so a fresh fork has
> nothing defined yet.


## 1. Pick a sink

Only names and types are visible; configuration and credentials never reach the notebook.

```python
sinks = mwlab.sinks()
sinks
```

```python
# Use the first enabled sink, or set a name explicitly.
SINK = next(s["name"] for s in sinks if s["enabled"])
SINK
```

## 2. Send search results

Attributes are sent with their event context. The sink's filters (IDS-only, types, tags) apply; pass
`apply_filters=False` to skip them. Its exclusions (e.g. `tlp:red`, warninglisted values) always apply.

```python
hits = mwlab.search_attributes(type="ip-dst", size=50)
mwlab.send_to_sink(SINK, attributes=hits)
```

## 3. Send your own records

Any `value` + `type`, with optional `category`, `comment`, `to_ids` and `tags`.

```python
mwlab.send_to_sink(SINK, records=[
    {"value": "192.0.2.10", "type": "ip-dst", "comment": "sent from a notebook", "tags": ["tlp:green"]},
])
```

Deliveries run on the sinks worker. Check the sink's status in **internals → sinks**, and every send in the audit log.
